
# Convergence Simulation Experiments

This notebook documents a series of simulation experiments exploring opinion dynamics and network convergence under different conditions. The experiments analyze how variables such as the number of agents, the presence and count of stubborn agents, edge failure probabilities, and network density affect the time it takes for a consensus to be reached.

### Overview of Experiments:
1. **Experiment 1 (No Stubborn Agents):** Investigates convergence behavior when all agents adjust their opinions freely.
2. **Experiment 2 (Single Stubborn Agent):** Explores the influence of one stubborn truth-seeking agent as the population scales.
3. **Experiment 3 & 4 (Scaling Stubborn Agents):** Measures convergence speed while increasing the number of stubborn agents in static populations ($N=20$ and $N=100$).
4. **Experiment 5 (Edge Failure Probability):** Examines the impact of communication failures on network consensus rates.
5. **Experiment 6 (Network Density):** Analyzes how connection density impacts opinion dissemination and convergence.
